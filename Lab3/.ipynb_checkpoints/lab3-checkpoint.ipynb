```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()
```

```python
df = pd.read_csv("student_exam_performance.csv")

df.head(-1)
```

```python
df.dtypes
```

```python
df.head().style.format({'exam_score': '{:.2f}%'})
```

```python
print(df.isna().sum())

df.duplicated()[df.duplicated()==True]
```

```python
print("Shape ( row, colimns): ", df.shape, "\n")

print("number of rows: ",df.shape[0])
print("number of columns: ",df.shape[1])
```

```python
df.describe(include='all')
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['exam_score'], bins=20, kde=True)
plt.title("Distribution of Exam Scores")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['exam_score'], bins=20, kde=True)
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Count")
plt.show()
```

```python
school_performance = df.groupby('school_type')['exam_score'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,5))
school_performance.plot(kind='bar' , color='skyblue')
plt.title("Average Exam Score by School Type")
plt.ylabel("Average Exam Score")
plt.show()

school_performance
```

```python
study_method_scores = df.groupby('study_method')['exam_score'].mean().sort_values(ascending=False)

study_method_scores.head(10)
```

```python
plt.figure( figsize = (8,5))
sns.scatterplot( x = 'study_hours_per_day' , y = 'exam_score' , data = df)
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Exam Score")
plt.show()
```

```python
plt.figure( figsize =(8,6))

numeric_cols= [ 'exam_score' , 'study_hours_per_day' ,'attendance_percentage' , 'assignment_completion_rate' , 'sleep_hours']

sns.heatmap(df[numeric_cols].corr() , annot=True , cmap='coolwarm' , fmt=".2f")

plt.title("Correlation Matrix of Student Performance Factors")

plt.show()
```
