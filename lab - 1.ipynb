```python
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]

for mark in marks:
    print(mark)

for mark in marks:
    if mark >= 60:
        print(mark, "Pass")
    else:
        print(mark, "Fail")

for i in range(len(marks)):
    if marks[i] >= 60:
        result = "Pass"
    else:
        result = "Fail"
    print("Study hours:", study_hours[i],
          "Mark:", marks[i], "Result:", result)
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]

n = len(marks)
print("Number of samples:", n)

total_hours = 0
for h in study_hours:
    total_hours = total_hours + h
print("Average study hours:", total_hours / n)

total_marks = 0
for m in marks:
    total_marks = total_marks + m
print("Average marks:", total_marks / n)

highest = marks[0]
lowest = marks[0]
for m in marks:
    if m > highest:
        highest = m
    if m < lowest:
        lowest = m
print("Highest mark:", highest)
print("Lowest mark:", lowest)

for i in range(n):
    print("Study hours:", study_hours[i], "Mark:", marks[i])

for i in range(n):
    if marks[i] >= 60:
        print("Study hours:", study_hours[i], "Mark:", marks[i], "Pass")
    else:
        print("Study hours:", study_hours[i], "Mark:", marks[i], "Fail")

count = 0
for m in marks:
    if m >= 60:
        count = count + 1
print("Number of passes:", count)
```
